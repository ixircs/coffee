```python
import pandas as pd

f = "../data/raw/CMO-Historical-Data-Monthly.xlsx"

# Ambil nama kolom dari baris 4 (satuan di baris 5, kita catat manual)
names = pd.read_excel(f, sheet_name="Monthly Prices", header=None,
                      skiprows=4, nrows=1).iloc[0]

# Ambil datanya mulai baris 6
df = pd.read_excel(f, sheet_name="Monthly Prices", header=None, skiprows=6)
df.columns = list(names)
df = df.rename(columns={df.columns[0]: "date"})

# Ambil kopi doang + ubah tanggal jadi format datetime
d = df[["date", "Coffee, Arabica", "Coffee, Robusta"]].dropna()
d["dt"] = pd.to_datetime(d["date"].str.replace("M", "-"), format="%Y-%m")
d = d.set_index("dt")
```
